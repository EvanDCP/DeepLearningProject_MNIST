# KERAS ET DEEP LEARNING

Le code suivant permet d'utiliser un réseau de neurone à convolution en utilisant la librairie KERAS.

```python
# Pour pouvoir fonctionner avec Keras il faut redimensionner le tableau en un tableau 4 dimensions

x_train = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1)
input_shape = (28, 28, 1)

# S'assurer que les valeurs sont flottantes afin que nous puissions obtenir des points décimaux après la division
# Normaliser les codes RVB en le divisant à la valeur RVB max.

x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255
```

```python
# Import des modules Keras
from keras.models import Sequential
from keras.layers import Dense, Conv2D, Dropout, Flatten, MaxPooling2D

from tensorflow import keras
from tensorflow.keras import layers

model = Sequential()
model.add(Conv2D(20, kernel_size=(5,5), input_shape= (28, 28, 1)))

# Sous-échantillonnage des images en entrée cela permet également au model de se consentrer sur les caractèristiques les plus importantes de nos images
model.add(MaxPooling2D(pool_size=(2, 2))) 

model.add(Flatten())

# Premier layer avec 100 neurones
model.add(Dense(100, activation=tf.nn.relu)) 

# Pour la regularization ( Chapitre 3 )
model.add(Dropout(0.2)) 

# Output layer avec 10 neurones en sortie ( Chapitre 3) model.add(Dense(10,activation=tf.nn.softmax)) 
model.add(Dense(10, activation=tf.nn.softmax)) 
```

```python
model.compile(optimizer='sgd', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
H = model.fit(x=x_train,y=y_train, epochs=10)
```

Le code suivant nous permet d'évaluer notre réseau de neurones sur la base de test une fois l'entrainement terminé

```python
model.evaluate(x_test, y_test)
```
